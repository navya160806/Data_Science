```python
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
df
```

```python
#age distribution
#histplot= single data
fig=sns.histplot(df,x='age')
fig.set_title("age distribution of titanic passaenger")
fig.set_xlabel("age")
fig.set_ylabel("frequency")
```

```python
#survival count based on ghender
#hue=used to show cateogary in  the data 
fig=sns.countplot(df,x='survived',hue='sex')
fig.set_title("survival based on gender ")
fig.set_xlabel("survived")
fig.set_ylabel("count")
```

```python
fig=sns.barplot(df,x='sex',y='fare',hue='class')
fig.set_title("average fare by passengers class and gender")
fig.set_xlabel("sex")
fig.set_ylabel("average fare")
```

```python
fig=sns.countplot(df,x='class',palette='coolwarm')
fig.set_title("count of passengers by class")
fig.set_xlabel("passenger class")
fig.set_ylabel("count")
```

```python
#summary graph

sns.pairplot(df[['age','fare','survived','pclass']],hue='survived',palette='terrain')
```
