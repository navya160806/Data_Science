```python
# ---------- Method Overloading ----------
class Calculator:
    def add(self, a, b=0, c=0):   # overloading using default arguments
        return a + b + c

# ---------- Method Overriding ----------
class AdvancedCalculator(Calculator):
    def add(self, a, b=0, c=0):   # overriding parent method
        print("Overridden add method")
        return a + b + c

# Object creation
calc = Calculator()
adv_calc = AdvancedCalculator()

# Overloading demonstration
print(calc.add(5))        # 5
print(calc.add(5, 10))    # 15
print(calc.add(5, 10, 15))# 30

# Overriding demonstration
print(adv_calc.add(2, 3)) # Overridden method
```
